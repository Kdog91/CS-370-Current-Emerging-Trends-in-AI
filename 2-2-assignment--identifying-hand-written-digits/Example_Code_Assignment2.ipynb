```python
from __future__ import print_function
import numpy as np
from keras.datasets import mnist
from keras.models import Sequential
#from keras.layers.code import Dense, Activation # From the book, code replaced below
from keras.layers import Dense, Activation, Input # syntax change due to version bump
from keras.optimizers import SGD
#from keras.utils import np_utils # From the book, code replaced below
from keras.utils import to_categorical # syntax change due to version bump
```

---

### Network and Training Parameters
<div class="alert alert-block alert-danger" style="color:black;">
<b>Warnings:</b> As a reminder, you may receive some errors here, these are standard errors that can be ignored as they are environment based.
</div>

```python
#NB_EPOCH = 200 # From the book, code replaced below
NB_EPOCH = 20 # syntax change to speed up quick look testing
BATCH_SIZE = 128
VERBOSE = 1
NB_CLASSES = 10 # number of outputs = number of digits
OPTIMIZER = SGD() # SGD optimizer, explained later in this chapter
N_HIDDEN = 128
VALIDATION_SPLIT = 0.2 # how much TRAIN is reserved for VALIDATION
```

---
### Spliting Data Into Train and Test Sets

```python
# data: shuffled and split between train and test sets
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

---
### Shaping

```python
# X_train is 60000 rows of 28x28 values --> reshaped in 60000 x 784

RESHAPED = 784

X_train = X_train.reshape(60000, RESHAPED)
X_test = X_test.reshape(10000, RESHAPED)
X_train = X_train.astype('float32')
X_test = X_test.astype('float32')
```

```python
# normalize
X_train /= 255
X_test /= 255

print(X_train.shape[0], 'train samples')
print(X_test.shape[0], 'test samples')
```

```python
# convert class vectors to binary class matrices

# Y_train = np_utils.to_categorical(y_train, NB_CLASSES) # From the book, code replaced below
Y_train = to_categorical(y_train, NB_CLASSES) # Updated for current version

#Y_test = np_utils.to_categorical(y_test, NB_CLASSES) # From the book, code replaced below
Y_test = to_categorical(y_test, NB_CLASSES) # Updated for current version
```

---
### Model Building/Definition

<div class="alert alert-block alert-danger" style="color:black;">
<b>Warnings:</b> As a reminder, you may receive some errors here, these are standard errors that can be ignored as they are environment based.
</div>

Potentially helpful links: 
- https://keras.io/api/models/sequential/
- https://keras.io/api/models/sequential/#add-method
- https://keras.io/api/layers/activations/
- https://keras.io/api/layers/
- https://keras.io/api/layers/core_layers/

```python
# 10 outputs
# final stage is softmax

model = Sequential()
#model.add(Dense(NB_CLASSES, input_shape=(RESHAPED,))) # From the book, code replaced below
model.add(Input(shape=(RESHAPED,))) # Updated for current version
model.add(Dense(NB_CLASSES)) # Updated for current version
model.add(Activation('softmax'))
model.summary()
```

### Model Compilation

Potentially helpful links: 
- https://keras.io/api/models/model_training_apis/#compile-method

```python
model.compile(loss='categorical_crossentropy', optimizer=OPTIMIZER, metrics=['accuracy'])
```

### Model Training
<div class="alert alert-block alert-danger" style="color:black;">
<b>Warnings:</b> As a reminder, you may receive some errors here, these are standard errors that can be ignored as they are environment based.
</div>

<div class="alert alert-block alert-danger" style="color:black;">

</div>

Potentially helpful links: 
- https://keras.io/api/models/model_training_apis/#fit-method

```python
history = model.fit(X_train, Y_train, batch_size=BATCH_SIZE, epochs=NB_EPOCH, verbose=VERBOSE, validation_split=VALIDATION_SPLIT)
```

### Model Evaluation
Potentially helpful links: 
- https://keras.io/api/models/model_training_apis/#evaluate-method

```python
score = model.evaluate(X_test, Y_test, verbose=VERBOSE)

print("Test Score:", score[0])
print("Test Accuracy:", score[1])
```
